```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "I order pizza from the food delivery app",
    "The delivery service provides fresh food",
    "I like ordering burgers and pizza online",
    "Food delivery apps offer fast service",
    "Fresh burgers are available for online orders"
]

print(documents)
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

bow_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("Bag of Words Representation:")
print(bow_df)
```

```python
unigram = CountVectorizer(ngram_range=(1,1))

unigram_features = unigram.fit_transform(documents)

print("Unigrams:")
print(unigram.get_feature_names_out())
```

```python
bigram = CountVectorizer(ngram_range=(2,2))

bigram_features = bigram.fit_transform(documents)

print("Bigrams:")
print(bigram.get_feature_names_out())
```

```python
trigram = CountVectorizer(ngram_range=(3,3))

trigram_features = trigram.fit_transform(documents)

print("Trigrams:")
print(trigram.get_feature_names_out())
```

```python
word_frequency = bow_df.sum(axis=0)

print("Word Frequency:")
print(word_frequency.sort_values(ascending=False))
```

```python
plt.figure(figsize=(10,5))

word_frequency.sort_values(ascending=False).head(10).plot(kind='bar')

plt.title("Top 10 Word Frequency Distribution")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)

plt.show()
```
